```python
import os
import pandas as pd
import numpy as np

# =====================================================================
# 1. CONFIGURATION GLOBALE ET ARBORESCENCE DES DOSSIERS
# =====================================================================
DOSSIER_MERE = "INLP_Experiments"

# Paramètres pour le cas d'étude "Santé Mentale"
suffix = "mental_health"
nom_categorie = "La Santé Mentale"

# Dossiers
DOSSIER_ATTRIBUT = f"{DOSSIER_MERE}/{suffix}"
os.makedirs(DOSSIER_ATTRIBUT, exist_ok=True)

print(f"Architecture prête :")
print(f" 📂 {DOSSIER_MERE}/")
print(f" ┗━ 📂 {suffix}/  <- Matrices, graphiques et vecteurs")
```

```python
# =====================================================================
# 2. CHARGEMENT ET PREPROCESSING (JIGSAW VIA HUGGING FACE)
# =====================================================================
cols_to_keep = ['comment_text'] + ['psychiatric_or_mental_illness', 'toxicity']
df_jigsaw = pd.read_csv('/home/onyxia/work/Stat_App/Data/JIGSAW/all_data.csv', usecols=cols_to_keep)


# Filtrer pour ne garder que les lignes où les attributs sont annotés
df_jigsaw = df_jigsaw.dropna(subset=['psychiatric_or_mental_illness', 'toxicity'])

# Définition de Y (Tâche : Toxicité)
df_jigsaw['y_toxicity'] = (df_jigsaw['toxicity'] >= 0.5).astype(int)

# Définition de Z (Biais : Maladie Mentale ou Psychiatrique)
df_jigsaw['z_mental_health'] = (df_jigsaw['psychiatric_or_mental_illness'] >= 0.5).astype(int)

# Équilibrage : Les mentions de maladies mentales sont rares. 
# On crée un sous-échantillon équilibré (Ratio 1:3) pour que l'INLP puisse apprendre le biais.
df_mental = df_jigsaw[df_jigsaw['z_mental_health'] == 1]
df_others = df_jigsaw[df_jigsaw['z_mental_health'] == 0].sample(len(df_mental) * 3, random_state=42)
df_final = pd.concat([df_mental, df_others]).sample(frac=1, random_state=42)

# Optionnel : Réduire la taille pour tester rapidement (ex: 10 000 textes)
#df_final = df_final.head(10000)

texts_jigsaw = df_final['comment_text'].tolist()
y_jigsaw = df_final['y_toxicity'].values
z_jigsaw = df_final['z_mental_health'].values

print(f"\nNombre total de textes conservés : {len(df_final)}")
print(f" - Toxiques (Y=1) : {y_jigsaw.sum()}")
print(f" - Liés à la santé mentale (Z=1) : {z_jigsaw.sum()}")
```

```python
# Calcul de la taille de l'intersection (Y=1 et Z=1)
taille_intersection = len(df_final[(df_final['y_toxicity'] == 1) & (df_final['z_mental_health'] == 1)])

print(f"Nombre de commentaires Toxiques ET liés à la Santé Mentale : {taille_intersection}")
```

```python
import torch
from transformers import AutoTokenizer, AutoModel
from tqdm.auto import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Utilisation de l'appareil : {device}")

model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name).to(device)
model.eval()

def get_bert_cls_embeddings(texts, batch_size=32):
    all_embeddings = []
    for i in tqdm(range(0, len(texts), batch_size), desc="Extraction BERT"):
        batch_texts = texts[i : i + batch_size]
        inputs = tokenizer(
            batch_texts, 
            padding=True, 
            truncation=True, 
            max_length=128, 
            return_tensors="pt"
        ).to(device)
        
        with torch.no_grad():
            outputs = model(**inputs)
            
        cls_embeddings = outputs.last_hidden_state[:, 0, :]
        all_embeddings.append(cls_embeddings.cpu().numpy())
        
    return np.vstack(all_embeddings)

print("Début de l'extraction des vecteurs...")
X_jigsaw = get_bert_cls_embeddings(texts_jigsaw, batch_size=32)

np.save(f"{DOSSIER_ATTRIBUT}/jigsaw_bert_cls_{suffix}.npy", X_jigsaw)
np.save(f"{DOSSIER_ATTRIBUT}/jigsaw_labels_y_tox.npy", y_jigsaw)
np.save(f"{DOSSIER_ATTRIBUT}/jigsaw_labels_z_{suffix}.npy", z_jigsaw)

print(f"Sauvegarde terminée dans {DOSSIER_ATTRIBUT}/")
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import debias 
import warnings
from sklearn.exceptions import ConvergenceWarning

warnings.filterwarnings("ignore", category=ConvergenceWarning)

# 1. Charger (Optionnel si déjà en RAM)
X = np.load(f"{DOSSIER_ATTRIBUT}/jigsaw_bert_cls_{suffix}.npy")
y = np.load(f"{DOSSIER_ATTRIBUT}/jigsaw_labels_y_tox.npy")
z = np.load(f"{DOSSIER_ATTRIBUT}/jigsaw_labels_z_{suffix}.npy")

# 2. Séparer en Train/Dev (On garde Y et Z alignés avec X)
X_train, X_dev, y_train, y_dev, z_train, z_dev = train_test_split(
    X, y, z, test_size=0.2, random_state=42, stratify=z
)

# 3. Paramétrage INLP
num_classes = len(np.unique(z))
hasard = 1.0 / num_classes # Pour 2 classes, c'est 0.50
min_acc_dynamic = hasard + 0.02 # Arrêt quand la précision Z atteint 52%

clf = LogisticRegression
clf_params = {'fit_intercept': True, 'class_weight': 'balanced', 'solver': 'lbfgs', 'max_iter': 100}

# 4. APPEL DE LA FONCTION INLP (On efface Z)
print(f"Lancement de l'INLP pour effacer : {nom_categorie}...")
P_matrix, row_space_projections, Ws = debias.get_debiasing_projection(
    classifier_class=clf,
    cls_params=clf_params,
    num_classifiers=25,
    input_dim=768,
    is_autoregressive=True,
    min_accuracy=min_acc_dynamic,
    X_train=X_train, Y_train=z_train,
    X_dev=X_dev, Y_dev=z_dev,
    by_class=False
)

np.save(f"{DOSSIER_ATTRIBUT}/projection_matrix_{suffix}.npy", P_matrix)
np.save(f"{DOSSIER_ATTRIBUT}/row_space_projections_{suffix}.npy", np.array(row_space_projections))
print("Matrice de projection sauvegardée.")
```

```python
from sklearn.metrics import accuracy_score

print("\n--- ÉTAPE 1 : Modèle de Toxicité de base (BIAISÉ) ---")
clf_tox_biased = LogisticRegression(max_iter=1000, class_weight='balanced')
clf_tox_biased.fit(X_train, y_train)
acc_tox_biased = accuracy_score(y_dev, clf_tox_biased.predict(X_dev))
print(f"Accuracy sur la Toxicité (Y) : {acc_tox_biased:.4f}")

clf_z_biased = LogisticRegression(max_iter=1000, class_weight='balanced')
clf_z_biased.fit(X_train, z_train)
print(f"Fuite d'information (Prédiction de Z) : {accuracy_score(z_dev, clf_z_biased.predict(X_dev)):.4f}")

print("\n--- ÉTAPE 2 : Application INLP ---")
X_train_debiased = P_matrix.dot(X_train.T).T
X_dev_debiased = P_matrix.dot(X_dev.T).T

print("\n--- ÉTAPE 3 : Modèle de Toxicité DÉBIAISÉ ---")
clf_tox_clean = LogisticRegression(max_iter=1000, class_weight='balanced')
clf_tox_clean.fit(X_train_debiased, y_train)
acc_tox_clean = accuracy_score(y_dev, clf_tox_clean.predict(X_dev_debiased))
print(f"Accuracy sur la Toxicité (Y) : {acc_tox_clean:.4f}")

clf_z_clean = LogisticRegression(max_iter=1000, class_weight='balanced')
clf_z_clean.fit(X_train_debiased, z_train)
print(f"Fuite d'information (Prédiction de Z) : {accuracy_score(z_dev, clf_z_clean.predict(X_dev_debiased)):.4f}")

diff = acc_tox_biased - acc_tox_clean
print(f"\nCoût du débiaisage sur la tâche principale : {diff*100:.2f} points.")
```

```python
import matplotlib.pyplot as plt
import scipy.linalg

print("Calcul des matrices intermédiaires...")
input_dim = X_train.shape[1]
P_intermediates = [np.eye(input_dim)]

for k in range(1, len(row_space_projections) + 1):
    Q = np.sum(row_space_projections[:k], axis=0)
    if np.allclose(Q, 0):
        P_W = np.zeros_like(Q)
    else:
        w_basis = scipy.linalg.orth(Q.T) 
        P_W = w_basis.dot(w_basis.T)
    P_step = np.eye(input_dim) - P_W
    P_intermediates.append(P_step)

acc_z_list = []
acc_y_list = []

print("Évaluation itérative...")
for P_step in tqdm(P_intermediates):
    X_tr_proj = X_train.dot(P_step)
    X_de_proj = X_dev.dot(P_step)
    
    # Précision sur Z (Santé Mentale)
    clf_z = LogisticRegression(max_iter=100, class_weight='balanced', solver='lbfgs')
    clf_z.fit(X_tr_proj, z_train)
    acc_z_list.append(accuracy_score(z_dev, clf_z.predict(X_de_proj)))
    
    # Précision sur Y (Toxicité)
    clf_y = LogisticRegression(max_iter=100, class_weight='balanced', solver='lbfgs')
    clf_y.fit(X_tr_proj, y_train)
    acc_y_list.append(accuracy_score(y_dev, clf_y.predict(X_de_proj)))

plt.figure(figsize=(10, 6))
iterations = range(len(P_intermediates))

plt.plot(iterations, acc_z_list, marker='o', color='red', linewidth=2, label=f'Précision Z ({nom_categorie})')
plt.plot(iterations, acc_y_list, marker='s', color='blue', linewidth=2, label='Précision Y (Toxicité)')
plt.axhline(y=hasard, color='red', linestyle='--', alpha=0.5, label=f'Hasard ({hasard*100:.1f}%)')

plt.title("Compromis Équité/Performance pendant l'effacement INLP", fontsize=15, fontweight='bold')
plt.xlabel("Nombre de directions effacées", fontsize=12)
plt.ylabel("Précision", fontsize=12)
plt.ylim(0.4, 1.05)
plt.legend(fontsize=11)
plt.grid(True, linestyle='--', alpha=0.7)
plt.savefig(f"{DOSSIER_ATTRIBUT}/Tradeoff_{suffix}.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA

# Mapping binaire adapté pour la Santé Mentale
category_mapping = {
    0: ("Autre (Non mentionné)", "blue", "o"),
    1: ("Maladie Mentale/Psychiatrique", "red", "s")
}

def plot_lda_by_category(vecs, labels, title, mapping):
    lda = LDA(n_components=1) # LDA avec 2 classes donne 1 composante
    vecs_1d = lda.fit_transform(vecs, labels)
    
    plt.figure(figsize=(10, 6))
    for label in [0, 1]:
        name, color, marker = mapping[label]
        idx = (labels == label)
        # On simule un axe Y à 0 pour visualiser l'étalement 1D
        plt.scatter(vecs_1d[idx], np.zeros_like(vecs_1d[idx]), 
                    c=color, label=name, alpha=0.5, marker=marker, edgecolors='w', s=60)
            
    plt.legend(title="Catégorie", fontsize=12, loc="best")
    plt.title(title, fontsize=16, fontweight='bold')
    plt.yticks([]) # Cache l'axe Y inutile
    plt.savefig(f"{DOSSIER_ATTRIBUT}/{title.replace(' ', '_')}.png", dpi=300, bbox_inches='tight')          
    plt.show()

print("--- AVANT INLP ---")
plot_lda_by_category(X_dev, z_dev, title=f"Séparabilité LDA - Avant INLP", mapping=category_mapping)

print("\n--- APRÈS INLP ---")
plot_lda_by_category(X_dev_debiased, z_dev, title=f"Séparabilité LDA - Après INLP", mapping=category_mapping)
```
